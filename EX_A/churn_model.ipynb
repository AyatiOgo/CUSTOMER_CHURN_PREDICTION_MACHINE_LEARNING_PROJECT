## Feature Engineering And Model Prediction


```python
df.info()
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

int_columns = ['TotalCharges', 'MonthlyCharges' , 'tenure' ]
cat_columns  = [
    'gender',
    'Partner',
    'Dependents',
    'PhoneService',
    'MultipleLines',
    'InternetService',
    'TechSupport',
    'StreamingMovies',
    'Contract',
    'PaperlessBilling',
    'PaymentMethod',
]

preprocessor  = ColumnTransformer(
                transformers=[
                    ('num', StandardScaler() , int_columns),
                    ('cat', OneHotEncoder(handle_unknown='ignore'), cat_columns)
                ]
)

model = Pipeline([
            ('preprocessor', preprocessor ),
            ('classifier', LinearRegression() ),
])

#traning my dataset 
cols = int_columns + cat_columns
X = df[cols]
y = df['Churn']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

model.fit(X_train, y_train)
```

#### Checking for Accuracy

```python
predictions = model.predict(X_test)
predictions_class = (predictions >= 0.5).astype(int)
accuracy = accuracy_score(y_test, predictions_class)

print(accuracy)
```

```python
print(classification_report(y_test, predictions_class))
```

```python
print(confusion_matrix(y_test, predictions_class))
```
